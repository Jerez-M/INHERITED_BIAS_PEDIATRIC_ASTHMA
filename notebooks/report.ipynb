## The nursing lab

One child, scored twice. Change the numbers and watch the verdict move.

```python
import pyspiro
H = pyspiro.HANKINSON_1999()
P = pyspiro.HANKINSON_1999.Parameters

SEX, AGE, HEIGHT_CM, FEV1_L = 1, 12, 150, 2.50      # 1 = male, 0 = female

for label, eth in [("scored as Caucasian", 1), ("scored as African-American", 2)]:
    z = H.zscore(sex=SEX, age=AGE, height=HEIGHT_CM, ethnicity=eth,
                 parameter=P.FEV1, value=FEV1_L)
    verdict = "below the lower limit of normal" if z < -1.645 else "normal"
    print(f"{label:30s} z = {z:+.2f}   {verdict}")
```

The child did not change. Only the label did.
